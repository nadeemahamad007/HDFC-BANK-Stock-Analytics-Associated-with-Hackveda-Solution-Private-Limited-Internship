# Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Step 2 : Load the dataset

```python
from pathlib import Path
file_path = Path("../data/HDFCBANK_Stock_Data.csv")
data = pd.read_csv(file_path, parse_dates=["Date"])
```

```python
# Display basic information
data.head()
```

# Step 3: Preprocess the data

```python
print("\nMissing Values: \n", data.isnull().sum())
```

```python
print("\nDataset Info:")
data.info()
```

```python
# Convert the date column to datetime
data['Date'] = pd.to_datetime(data['Date'])
```

```python
# Set the date column as the index
data.set_index('Date', inplace=True)
```

# Step 4: Plot the closing price over time

```python
plt.figure(figsize=(12, 6))
plt.plot(data.index, data['Close'], label='Closing Price', color='blue')
plt.title('HDFCBANK Closing Price Over Time')
plt.xlabel('Date')
plt.ylabel('Closing Price (INR)')
plt.legend()
plt.grid()
plt.show()
```

# Step 4: Identify the overall trend 
# (e.g., uptrend or downtrend)

```python
# Calculate a simple linear regression for the trend line
x = np.arange(len(data))  # Time indices
y = data['Close'].values
```

```python
# Fit a linear regression line
coefficients = np.polyfit(x, y, 1)  # Degree 1 for a straight line
trend_line = np.poly1d(coefficients)
```

```python
# Add the trend line to the plot
plt.figure(figsize=(12, 6))
plt.plot(data.index, data['Close'], label='Closing Price', color='blue')
plt.plot(data.index, trend_line(x), label='Trend Line', color='red', linestyle='--')
plt.title('HDFCBANK Trend Line Over Time')
plt.xlabel('Date')
plt.ylabel('Closing Price (INR)')
plt.legend()
plt.grid()
plt.show()
```

# Step 5: Validate the trend line (draw parallel lines for the channel)

```python
# For simplicity, let's calculate the maximum deviation from the trend line
residuals = y - trend_line(x)
upper_channel = trend_line(x) + np.max(residuals)
lower_channel = trend_line(x) + np.min(residuals)
```

```python
# Plot the trend line with the channel
plt.figure(figsize=(12, 6))
plt.plot(data.index, data['Close'], label='Closing Price', color='blue')
plt.plot(data.index, trend_line(x), label='Trend Line', color='red', linestyle='--')
plt.plot(data.index, upper_channel, label='Upper Channel', color='green', linestyle=':')
plt.plot(data.index, lower_channel, label='Lower Channel', color='orange', linestyle=':')
plt.title('HDFCBANK Trend Line with Channels')
plt.xlabel('Date')
plt.ylabel('Closing Price (INR)')
plt.legend()
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(data.index, data['Close'], label='Closing Price', color='blue')
plt.plot(data.index, trend_line(x), label='Trend Line', color='red', linestyle='--')
plt.plot(data.index, upper_channel, label='Upper Channel', color='green', linestyle=':')
plt.plot(data.index, lower_channel, label='Lower Channel', color='orange', linestyle=':')
```

# Step 6: Annotate the chart

```python
plt.figure(figsize=(12, 6))
plt.plot(data.index, data['Close'], label='Closing Price', color='blue')
plt.plot(data.index, trend_line(x), label='Trend Line', color='red', linestyle='--')
plt.plot(data.index, upper_channel, label='Upper Channel', color='green', linestyle=':')
plt.plot(data.index, lower_channel, label='Lower Channel', color='orange', linestyle=':')
plt.annotate('Uptrend', xy=(data.index[len(data)//4], trend_line(len(data)//4)), 
             xytext=(data.index[len(data)//4], trend_line(len(data)//4) + 50),
             arrowprops=dict(facecolor='black', arrowstyle='->'))
plt.annotate('Channel Top', xy=(data.index[len(data)//2], upper_channel[len(data)//2]),
             xytext=(data.index[len(data)//2], upper_channel[len(data)//2] + 50),
             arrowprops=dict(facecolor='green', arrowstyle='->'))
plt.annotate('Channel Bottom', xy=(data.index[len(data)//2], lower_channel[len(data)//2]),
             xytext=(data.index[len(data)//2], lower_channel[len(data)//2] - 50),
             arrowprops=dict(facecolor='orange', arrowstyle='->'))

plt.title('HDFCBANK Trend Line with Annotations')
plt.xlabel('Date')
plt.ylabel('Closing Price (INR)')
plt.legend()
plt.grid()
plt.show()
```

# Step 8: Doing Averages work
# '50daysMA' : 'Green', '200daysMA': 'Red'

```python
# Calculate moving averages
data['MA50_Observations'] = data['Close'].rolling(window=50).mean()
data['MA200_Observations'] = data['Close'].rolling(window=200).mean()
```

```python
#function to plot the closing price along with specified moving averages
def plot_moving_averages(data, title, ma_columns):
    plt.figure(figsize=(14, 7))
    plt.plot(data.index, data['Close'], label='Closing Price', color='blue', alpha=0.7)
    for col, color in ma_columns.items():
        plt.plot(data.index, data[col], label=f'{col}', color=color, linewidth=1.5)
    plt.title(title)
    plt.xlabel('Date')
    plt.ylabel('Price')
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
# Plot 50-day and 200-day moving averages
plot_moving_averages(
    data,
    'HDFC BANK Closing Price with Moving Averages',
    {'MA50_Observations': 'green', 'MA200_Observations': 'red'}
)
```

```python
# Create trend masks
data['UpwardTrend'] = data['MA50_Observations'] > data['MA200_Observations']
data['DownwardTrend'] = data['MA200_Observations'] > data['MA50_Observations']

# Plot with shaded trends
plt.figure(figsize=(14, 7))
plt.plot(data.index, data['Close'], label='Closing Price', color='blue', alpha=0.7)
plt.plot(data.index, data['MA50_Observations'], label='50-observation MA', color='green', linewidth=1.5)
plt.plot(data.index, data['MA200_Observations'], label='200-observation MA', color='red', linewidth=1.5)
```

```python
# Fill between for Upward Trend
plt.fill_between(
    data.index,
    data['MA50_Observations'],
    data['MA200_Observations'],
    where=data['UpwardTrend'],
    color='lightgreen',
    alpha=0.8,
    label='Upward Trend'
)
```

```python
# Fill between for Downward Trend
plt.fill_between(
    data.index,
    data['MA50_Observations'],
    data['MA200_Observations'],
    where=data['DownwardTrend'],
    color='lightcoral',
    alpha=0.4,
    label='Downward Trend'
)

plt.title('HDFC BANK Closing Price with Trends Highlighted')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.grid(True)
plt.show()
```
